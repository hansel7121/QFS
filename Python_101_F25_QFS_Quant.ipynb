### Loops
As with the conditional operation, this is a key skill. In particular, it allows for us to scale up the number of operations without having to constantly write code to perform each one.

A for loop iterates over a sequence, for instance a list, string, dictionary,...

````python
x = [1,2,3,4,5]
for i in x:
  print(i)
````

````python
for i in range(5):
  print(i,x[i])
````

````python
word = 'hello world'
for letter in word:
    print(letter)
````

**Nested For Loops** - for loops within for loops

````python
count = 0
for i in range(10):
  for j in range(5):
    count += 1

# what does count equal?
print(count)
````

**While loops** - execute when the condition is true, and break when the condition becomes false.

````python
count = 10
sum = 0

while count > 0:
  sum += count
  print(sum)
  count -= 1
````

**Exercise**: Create a function called pigLatin that takes a string word and translates it to pig latin. The rules for pig latin are as follows:
* If the first letter is a consonant, move the first consonants to the end and add an "-ay" sound. For instance, pig latin translates to ig-pay atin-lay.
  * A better way to interpret this rule (particularly for consonant clusters) is to take the first cluster in the word, before the first vowel. For instance, "smile" becomes "ile-smay"
* If the first letter is a vowel, add "-way" to the end instead, like above translates to above-way.

````python
def pigLatin(word):
  vowels = ['a','e','i','o','u']
  if word[0] in vowels:
    word += 'way'
    print(word)
  else:
    word += word[0]
    word = word[1:]
    word += 'ay'
    print(word)

pigLatin('brian')
pigLatin('apple')
````
